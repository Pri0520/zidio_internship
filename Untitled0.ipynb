```python
code = '''
import streamlit as st
import pandas as pd
import plotly.express as px

st.set_page_config(page_title="Stock Forecast Dashboard", layout="wide")
st.title("📈 Stock Forecast Dashboard")
st.markdown("Upload your CSVs from ARIMA, Prophet, and LSTM models to visualize the forecast.")

uploaded_file = st.file_uploader("📁 Upload a forecast CSV file", type=["csv"])

if uploaded_file is not None:
    df = pd.read_csv(uploaded_file)
    st.success("✅ File uploaded successfully!")

    model_column = df.columns[1]
    model_name = model_column.replace("_Forecast", "")

    st.subheader(f"{model_name} Forecast (Next 30 Days)")
    fig = px.line(df, x="Date", y=model_column, title=f"{model_name} Forecast")
    st.plotly_chart(fig, use_container_width=True)

    st.subheader("📋 Forecast Table")
    st.dataframe(df.tail(30), use_container_width=True)

    csv = df.to_csv(index=False).encode('utf-8')
    st.download_button(label="⬇️ Download Forecast CSV", data=csv, file_name=f"{model_name}_forecast.csv", mime='text/csv')
'''

# Save the Streamlit app to file
with open("app.py", "w") as f:
    f.write(code)

print("✅ app.py created successfully")
```
